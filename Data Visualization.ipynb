```python
# Dictionary with multiple series
data = {
    'x': [1, 2, 3, 4, 5],
    'series1': [1, 4, 9, 16, 25],
    'series2': [2, 3, 5, 7, 11],
    'series3': [5, 7, 6, 8, 7]
}

# Line format for each series
formats = {
    'series1': 'r-o',   # red line, circle markers
    'series2': 'g--s',  # green dashed, square markers
    'series3': 'b-.^'   # blue dash-dot, triangle-up markers
}

# Loop through series and plot
for key, fmt in formats.items():
    plt.plot('x', key, fmt, data=data, label=key)

plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.title("Multiple Series from Dict")
plt.legend()
plt.grid(True)
plt.show()
```

```python

# Sample data: 4 different Y series
x = np.linspace(0, 10, 100)
y_data = [
    np.sin(x),
    np.cos(x),
    np.tan(x),
    np.exp(-x)
]

colors = ['red', 'green', 'blue', 'purple']
titles = ['Sine', 'Cosine', 'Tangent', 'Exponential Decay']

# Create 2x2 grid of subplots
fig, axss = plt.subplots(2, 2, figsize=(10, 6))

# Flatten axs to make it 1D so we can use indexing
axs_flat = axss.flatten()

# Loop through each subplot and assign data, color, title
for i in range(4):
    axs_flat[i].plot(x, y_data[i], color=colors[i])
    axs_flat[i].set_title(titles[i])
    axs_flat[i].grid(True)

# Adjust layout to prevent overlap
plt.tight_layout()
plt.show()
```

```python
#Functions Gallerey using flatten
#_import fn
import matplotlib.pyplot as plt
import numpy as np

#_gallery
functions=[
    (np.sin,'sin fn','red'),
    (np.cos,'cosin fn','blue'),
    (np.tan,'tangent fn','purple'),
    (np.sqrt,'squrt fn','green'),
    (np.exp,'exponential fn','yellow'),
    (np.log,'log fn','brown')
]

#_Creating subplots
fig ,axs = plt.subplots(2,3,figsize=(12,6))
flattened_axs=axs.flatten()

#_looping through subplots
for index, (fn,label,color) in enumerate(functions):
    if label=='squrt fn' or label=='log fn':
        if label=='log fn':
            X=np.arange(0.1,10,0.01)
        else : X=np.arange(0,10,0.01)
    elif label=='tangent fn':
        X=np.array(np.arange(-2*np.pi,np.pi*2,np.pi/51))
    else:
        X=np.arange(-10,10,0.01)
    Y=fn(X)
    flattened_axs[index].plot(X,Y,color=color,linewidth=2)
    flattened_axs[index].set_title(label)
    flattened_axs[index].grid(True,color='crimson')

#_view subplots
plt.tight_layout()
plt.show()
    
```

```python
#Bar Chart
categories = ['Electronics', 'Clothing', 'Books']
revenue = [1500, 1200, 900]
bar= plt.bar(categories,revenue)
for  index,rect in enumerate(bar):
    plt.text(categories[index] ,  # x position (center of bar)
             revenue[index],                   # y position (top of bar)
             revenue[index],              # label text
             ha='center', va='bottom')
plt.show()
```
